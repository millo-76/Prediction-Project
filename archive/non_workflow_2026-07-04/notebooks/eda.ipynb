# Exploratory Data Analysis

> Goal: inspect, clean, and identify model-ready columns from the 2026 Oracle's Elixir LoL dataset.

## 1) Setup and Display Options

```python
import pandas as pd

# Show every column in notebook outputs (avoid "..." truncation).
pd.set_option('display.max_columns', None)
```

## 2) Load Raw CSV

```python
# Load raw match-level and player/team-level data.
df = pd.read_csv('../data/raw/2026_LoL_OraclesElixir.csv', low_memory=False)

# Quick preview to confirm file loaded and inspect sample rows.
df.head()
```

## 3) Identify Columns and Profile Raw Data

```python
# List every column name so we can choose modeling features explicitly.
print(df.columns.tolist())
```

```python
# Check row count, non-null counts, and dtypes for cleaning decisions.
df.info()
```

```python
# Summary stats for numeric and categorical fields.
df.describe(include='all')
```

## 4) Filter to Complete Rows and Keep Core Columns

Core columns selected for the team-level prediction dataset:

- gameid
- date
- patch
- league
- side
- teamname
- result

```python
# Keep only rows marked complete in the source export.
df = df[df["datacompleteness"] == "complete"]

# Restrict to columns needed for modeling / downstream processing.
cols = ["gameid", "date", "patch", "league", "side", "teamname", "result"]
df[cols].head()
```

## 5) Quality Checks on Selected Fields

```python
# Confirm side labels are balanced/expected (e.g., Blue vs Red).
df["side"].value_counts()
```

```python
# Check target distribution (win/loss encoded in result).
df["result"].value_counts()
```

```python
# Count unique team names and preview a sample to spot naming inconsistencies.
unique_team_count = df["teamname"].nunique()
teamname_preview = df["teamname"].head()

unique_team_count, teamname_preview
```

```python
# Inspect repeated game IDs to verify expected row granularity.
df["gameid"].value_counts().head()
```

```python
# Review available positions before filtering to team-level rows.
df["position"].value_counts()
```

## 6) Final Team-Level Filter for Modeling

```python
# Keep only team-level rows (one row per team per game context).
df = df[df["position"] == 'team']

# Re-check game ID counts after filtering.
df["gameid"].value_counts().head()
```

## 7) Standardize the Case of the Sides Column.

```python
df["side"] = df["side"].str.lower() # Standardize side labels to lowercase for consistency.
df["side"].value_counts()
```

```python
df[df["position"] == 'team']    # Show team-level rows to confirm filtering worked as intended.
```

## 8) Cleaning Summary (Before/After Shapes)

This section documents how row counts changed at each cleaning stage.

```python
# Recreate key row-count checkpoints for a documented cleaning summary.
raw_df = pd.read_csv('../data/raw/2026_LoL_OraclesElixir.csv', low_memory=False)
complete_df = raw_df[raw_df["datacompleteness"] == "complete"]
team_df = complete_df[complete_df["position"] == "team"]

summary = pd.DataFrame(
    {
        "step": [
            "Raw CSV loaded",
            "After datacompleteness == 'complete'",
            "After position == 'team'",
        ],
        "rows": [len(raw_df), len(complete_df), len(team_df)],
        "unique_gameids": [
            raw_df["gameid"].nunique(),
            complete_df["gameid"].nunique(),
            team_df["gameid"].nunique(),
        ],
    }
)

# Show step-to-step changes; negative values indicate rows/IDs removed.
summary["rows_change_vs_prev_step"] = summary["rows"].diff().fillna(0).astype(int)
summary["unique_gameids_change_vs_prev_step"] = summary["unique_gameids"].diff().fillna(0).astype(int)

summary
```
